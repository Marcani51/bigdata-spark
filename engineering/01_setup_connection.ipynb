# 01 — Setup & Connection TestNotebook ini memverifikasi koneksi PySpark ke Spark cluster.**Prasyarat:** Docker Compose sudah running (`docker compose up -d`)

## 1.1 Inisialisasi SparkSession

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("01_SetupTest") \
    .master("spark://spark-master:7077") \
    .config("spark.executor.memory", "1g") \
    .config("spark.driver.memory", "1g") \
    .getOrCreate()

print(f"✅ Spark Version : {spark.version}")
print(f"✅ Spark Master  : {spark.sparkContext.master}")
print(f"✅ App Name      : {spark.sparkContext.appName}")
```

## 1.2 Test Baca Dataset

```python
# Quick test — baca CSV
df_test = spark.read.csv("/data/retail_sales_dataset.csv", header=True, inferSchema=True)

print(f"✅ Total Rows   : {df_test.count()}")
print(f"✅ Total Columns: {len(df_test.columns)}")
print(f"✅ Columns      : {df_test.columns}")
df_test.show(5)
```

## 1.3 Test Spark SQL

```python
df_test.createOrReplaceTempView("test_table")

result = spark.sql("SELECT COUNT(*) as total FROM test_table")
result.show()
print("✅ Spark SQL berfungsi dengan baik!")
```

## 1.4 Cleanup

```python
spark.stop()
print("✅ SparkSession ditutup. Setup berhasil!")
```
